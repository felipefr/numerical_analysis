```python
import numpy as np
import matplotlib.pyplot as plt

N = 20;

x = np.logspace(1,N,N)**-1; 
print(x)

y1 = ((x+1)**2-1)/x
y2 = x + 2

plt.plot(x,y1,'o')
plt.plot(x,y2,'r+')
plt.legend(['y1', 'y2'])
plt.xscale('log')
plt.yscale('log')
plt.grid()
```
